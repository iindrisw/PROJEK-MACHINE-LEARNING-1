#Perlakuan data-Deteksi dan penanganan pencilan dengan metode IQR, lalu penskalaan


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# 1. Load data asli
df_clean = pd.read_csv('student-por.csv', sep=';')

# Identifikasi kolom numerik
num_cols = df_clean.select_dtypes(include=['number']).columns.tolist()
print("Fitur numerik sebelum penanganan:", num_cols)

# 2. Deteksi dan Penanganan Pencilan dengan Metode IQR
# Kita akan menerapkan IQR pada kolom numerik kontinu/memiliki rentang nilai lebar seperti 'absences' dan 'age'.
# Catatan: Nilai G1, G2, G3 adalah nilai ujian terstruktur sehingga nilai 0 atau batas bawah seringkali bermakna akademis,
# namun untuk demonstrasi pembersihan pencilan, kita akan fokus pada kolom 'absences' yang rawan outlier ekstrem.

for col in ['absences', 'age']:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Menghitung jumlah outlier
    outliers = df_clean[(df_clean[col] < lower_bound) | (df_clean[col] > upper_bound)]
    print(f"Jumlah outlier pada fitur '{col}': {len(outliers)}")

    # Mengganti nilai di luar batas dengan nilai batas terdekat (Capping)
    df_clean[col] = np.where(df_clean[col] < lower_bound, lower_bound, df_clean[col])
    df_clean[col] = np.where(df_clean[col] > upper_bound, upper_bound, df_clean[col])

print("\nPenanganan pencilan dengan IQR (capping) selesai.")
```

```python
# 3. Penskalaan Fitur (Scaling) menggunakan StandardScaler
scaler = StandardScaler()

# Kolom target G3 biasanya tidak diskalakan jika ingin mempertahankan interpretasi nilai asli,
# tetapi fitur input numerik lainnya perlu discaling.
features_to_scale = [col for col in num_cols if col != 'G3']

# Duplikat dataframe untuk menyimpan hasil scaling
df_scaled = df_clean.copy()
df_scaled[features_to_scale] = scaler.fit_transform(df_clean[features_to_scale])

print("=== CONTOH FITUR SETELAH DI-SCALING (StandardScaler) ===")
display(df_scaled[features_to_scale].head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Memuat kembali data asli tanpa modifikasi untuk visualisasi perbandingan
df_original = pd.read_csv('student-por.csv', sep=';')

# Membuat plot perbandingan
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# 1. Absences sebelum penanganan
sns.boxplot(ax=axes[0, 0], x=df_original['absences'], color='salmon')
axes[0, 0].set_title('Absences Sebelum Penanganan (Banyak Outlier)')
axes[0, 0].set_xlabel('Jumlah Absensi')

# 2. Absences sesudah penanganan dengan IQR Capping
sns.boxplot(ax=axes[0, 1], x=df_clean['absences'], color='lightgreen')
axes[0, 1].set_title('Absences Setelah IQR Capping (Batas Atas disesuaikan)')
axes[0, 1].set_xlabel('Jumlah Absensi')

# 3. Age sebelum penanganan
sns.boxplot(ax=axes[1, 0], x=df_original['age'], color='salmon')
axes[1, 0].set_title('Age Sebelum Penanganan')
axes[1, 0].set_xlabel('Umur')

# 4. Age sesudah penanganan dengan IQR Capping
sns.boxplot(ax=axes[1, 1], x=df_clean['age'], color='lightgreen')
axes[1, 1].set_title('Age Setelah IQR Capping')
axes[1, 1].set_xlabel('Umur')

plt.tight_layout()
plt.show()
```
