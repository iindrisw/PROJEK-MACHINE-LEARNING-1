```python
import pandas as pd

# 1. Load data (ingat pakai pemisah titik koma!)
df = pd.read_csv('student-por.csv', sep=';')

# 2. Tampilin daftar semua kolom beserta tipe datanya (int64 = angka, object = kategori/teks)
print("=== INFO TIPE DATA KOLOM ===")
df.info()

print("\n=== CONTOH ISI 5 BARIS PERTAMA ===")
display(df.head())

print("\n=== JUMLAH NILAI UNIK TIAP KOLOM (Biar gampang nentuin ini kategori atau bukan) ===")
print(df.nunique())
```

```python
print("\n=== CEK NILAI KOSONG (NULL) ===")
print(df.isnull().sum())
print("Total nilai kosong seluruh dataset:", df.isnull().sum().sum())

print("\n=== CEK BARIS DUPLIKAT ===")
print("Jumlah baris duplikat:", df.duplicated().sum())

# Jebakan khusus 5C-1: Cek siswa yang G3 (nilai akhir) nya 0 (absen ujian)
siswa_nol_g3 = (df['G3'] == 0).sum()
print(f"Jumlah siswa dengan nilai akhir G3 bernilai 0: {siswa_nol_g3}")
```

```python
print("\n=== STATISTIK DESKRIPTIF TARGET (G3) ===")
print(df['G3'].describe())

# Cek korelasi fitur numerik terhadap target G3 untuk mendeteksi data leakage G1 & G2
print("\n=== KORELASI FITUR TERHADAP TARGET (G3) ===")
num_cols = df.select_dtypes(include=['number'])
korelasi_target = num_cols.corr()['G3'].sort_values(ascending=False)
print(korelasi_target)
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['G3'], bins=20, kde=True, color='skyblue')
plt.title('Sebaran Nilai Akhir Siswa (G3)')
plt.xlabel('Nilai G3')
plt.ylabel('Frekuensi')
plt.show()
```
