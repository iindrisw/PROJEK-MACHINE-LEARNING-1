```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# 1. Load dataset (pastikan file student-por.csv sudah diunggah di Google Colab)
# Catatan: dataset ini menggunakan pemisah titik koma (;)
df = pd.read_csv('student-por.csv', sep=';')

# 2. Memeriksa jumlah baris dan kolom (df.shape)
print("=== SHAPE DATASET ===")
print("Jumlah baris dan kolom:", df.shape)

# 3. Memeriksa jumlah nilai kosong (df.isnull().sum())
print("\n=== NILAI KOSONG ===")
print(df.isnull().sum())
print("Total nilai kosong seluruh dataset:", df.isnull().sum().sum())

# 4. Memeriksa jumlah baris duplikat (df.duplicated().sum())
print("\n=== BARIS DUPLIKAT ===")
print("Jumlah baris duplikat:", df.duplicated().sum())
```
