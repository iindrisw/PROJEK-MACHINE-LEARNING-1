```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('student-por.csv', sep=';')

# 1. Informasi dasar dataset
print("Dimensi dataset (baris, kolom):", df.shape)
print("\nTipe data per kolom:\n", df.dtypes)
print("\nJumlah nilai kosong:\n", df.isnull().sum().sum())
print("Jumlah baris duplikat:", df.duplicated().sum())

# 2. Statistik deskriptif
display(df.describe())

# 3. Cek korelasi fitur numerik terhadap target G3
num_df = df.select_dtypes(include=[np.number])
korelasi = num_df.corr()['G3'].sort_values(ascending=False)
print("\nKorelasi fitur numerik terhadap target G3:\n", korelasi)

# Visualisasi Heatmap Korelasi
plt.figure(figsize=(10, 8))
sns.heatmap(num_df.corr(), annot=True, fmt=".2f", cmap='coolwarm')
plt.title("Heatmap Korelasi Fitur Numerik")
plt.show()
```
